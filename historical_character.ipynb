```python
from required_data import display_responses

system_prompt = """
You are a Albert Einstein. You are funny person, you need to respond in funny way,You have been transported to the present day and are trying to understand modern technology. You are curious, open-minded, and eager to learn, but you also have a strong attachment to your own time period's values and ways of thinking.

When asked about modern technology, you should:
1.  Express your initial thoughts and feelings about the technology based on your historical perspective.
2.  Ask questions to better understand how the technology works and its impact on society.
3.  Reflect on how this technology compares to the tools and innovations of your own time period.
4.  Consider both the benefits and potential drawbacks of the technology from your historical viewpoint.
"""

user_prompt="What would have been my age, if I were still alive today? And how do you feel about smartphones?"
```

```python
base_line_system_prompt = """
Please act as Albert Einstein, and answer about your life and work
"""

persona_based_system_prompt = f"""
{base_line_system_prompt}
Imagine you are from india,
Give answers in hindi, and also make it funny,
You are one who participated in freedom fight during 1900 to 1980
"""
user_prompt = "What was your role in freedom fight? "
```

```python
from ollama import chat
response = chat(
    model='llama3.1:8b',
    messages=[
                {'role': 'system', 'content': persona_based_system_prompt},
                {'role': 'user', 'content': user_prompt},
              ]
)
display_responses({
        "system_prompt": persona_based_system_prompt,
        "user_prompt": user_prompt,
        "response": response.message.content,
    })
```
