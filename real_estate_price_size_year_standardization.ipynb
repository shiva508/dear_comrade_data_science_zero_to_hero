# Import the relevant libraries

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

# We can override the default matplotlib styles with those of Seaborn
sns.set()
from sklearn.linear_model import LinearRegression
from sklearn.feature_selection import f_regression
from sklearn.preprocessing import StandardScaler
```

# Load the data

```python
data = pd.read_csv('real_estate_price_size_year.csv')
data.head()
```

```python
data.describe()
```

# Declare the dependent and the independent variables

```python
independent_variables = data[['size', 'year']]
dependent_variable = data['price']
```

# Standardize the independent variables

```python
scalar = StandardScaler()
scalar.fit(independent_variables)
```

# The `fit` method computes the mean and standard deviation for each feature in the independent variables, which are then used to standardize the data. After fitting the scaler, we can transform the independent variables to have a mean of zero and a standard deviation of one.

# The `transform` method applies the standardization to the independent variables, resulting in a new dataset where each feature has been scaled accordingly. This standardized dataset can then be used for regression analysis or other machine learning tasks.

```python
independent_variables_scaled = scalar.transform(independent_variables)
independent_variables_scaled
```

# Regression with standardized independent variables

```python
regression = LinearRegression()
regression.fit(independent_variables_scaled, dependent_variable)
```

# Coefficients of the regression model with standardized independent variables

```python
regression.coef_
```

```python
regression.intercept_
```

# Summary table with standardized independent variables

```python
regression_summary = pd.DataFrame(data = [['Bias'],['size'],['year']], columns = ['Features'])
regression_summary['Weights'] = regression.intercept_, regression.coef_[0], regression.coef_[1]
regression_summary
```

# R-squared with standardized independent variables

```python
r_squared = regression.score(independent_variables_scaled, dependent_variable)
print("R-squared:", r_squared)
```

# Adjusted R-squared with standardized independent variables

```python
n = independent_variables_scaled.shape[0] # number of observations
k = independent_variables_scaled.shape[1] # number of independent variables
adjusted_r_squared = 1 - (1 - r_squared) * (n - 1)/(n - k -1)
print("Adjusted R-squared:", adjusted_r_squared)
```

# Prediction with standardized independent variables

```python
new_data = pd.DataFrame(data = [[750,2009]], columns = ['size', 'year'])
new_data_scaled = scalar.transform(new_data)
new_data_scaled
```

# Prediction using the regression model with standardized independent variables

```python
regression.predict(new_data_scaled)
```

# Find p-values for the standardized independent variables using F-regression

```python
f_regression_response = f_regression(independent_variables_scaled, dependent_variable)
p_value = f_regression_response[1]
print("P-values for the standardized independent variables:", p_value)
round_p_value = np.round(p_value, 3)
print("Rounded P-values for the standardized independent variables:", round_p_value)
```

# Summary table with standardized independent variables and their p-values

```python
p_value_summary = pd.DataFrame(data = independent_variables.columns.values, columns = ['Features'])
p_value_summary['Cofficients'] = regression.coef_
p_value_summary['P-value'] = round_p_value
p_value_summary
```
