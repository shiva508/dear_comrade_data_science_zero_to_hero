```python
from enum import Enum

class Color(Enum):
    RED = "Red"
    GREEN = "Green"
    BLUE = "Blue"
    YELLOW = "Yellow"
    WHITE = "White"
    GRAY = "Gray"
    BLACK = "Black"
    ORANGE = "Orange"
    PURPLE = "Purple"

def get_signal_description(color:Color):
    match color:
        case Color.RED:
            return "Stop"
        case Color.GREEN:
            return "Go"
        case Color.YELLOW:
            return "Caution"
        case _:
            return "Unknown signal"

signal = get_signal_description(Color.RED)
print(signal)
```
