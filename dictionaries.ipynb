```python
user_info = {'name': 'Alice', 'age': 30, 'city': 'New York', 'country': 'US'}
print(user_info)
print(user_info['name'])  # Output: Alice
print(user_info['age'])   # Output: 30
print(user_info['city'])  # Output: New York
## print(user_info['address'])  # Output: KeyError: 'address'
print(user_info.get('address', 'Address not found'))  # Output: Address not found
del user_info['country']
print(f"After deleting country: {user_info}")
```
