```python
from ollama import Client, chat

from required_data import display_responses

client = Client(
  host='http://localhost:11434',
  headers={'x-some-header': 'some-value'}
)
```

```python
def get_completion(system_prompt, user_prompt):
    messages = [
      {
          "role": "system", "content": system_prompt
      },
      {
          "role": "user", "content": user_prompt
      }
    ]
    client_response = client.chat(model='qwen3.5', messages=messages)
    return client_response.message.content
```

```python
#system_prompt = "You are a helpful assistant."
# Adding role
system_prompt = "You are a certified professional organizer."
constraint_system_prompt = f"{system_prompt}. I have only 15 minutes.  a 20 dolor budget and limited floor space: I want to keep sentimental items, but maximize surface space"
user_prompt = "Please give me simple plan to decluster and organize my work."
response = get_completion(system_prompt, user_prompt)
display_responses({
        "system_prompt": system_prompt,
        "user_prompt": user_prompt,
        "response": response,
    })
```
