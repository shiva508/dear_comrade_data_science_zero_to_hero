# Function Information
## it is defined with keywords like def, lambda
## it can have parameters and return values
## it can be called/invoked to perform a task

```python
def welcome_message():
    """This function prints a welcome message to the user."""
    return f"Welcome, SHIVA!"
```

```python
message = welcome_message()
print(message)
```

# Function with Parameters
## Functions can accept parameters to perform tasks based on input values.
## Parameters are defined within the parentheses in the function definition.
## When calling the function, you provide arguments that correspond to these parameters.

```python
def greet_user(name):
    """This function greets the user with their name."""
    return f"Hello, {name}!"
```

```python
user_name = greet_user("Shiva")
print(user_name)
```

```python
def exponentiation_exp_2(x):
    return x ** 2

result = exponentiation_exp_2(2)
print(f"Raised to the power of 2:{result}")
```
