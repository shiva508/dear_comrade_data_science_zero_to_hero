# # Clustering Categorical Data

## Import the relevant libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
from sklearn.cluster import KMeans
```

# Load the data

```python
row_data = pd.read_csv('country_clusters.csv')
```

# Map the categorical variable to numerical values

```python
mapped_data = row_data.copy()
mapped_data["Language"] = mapped_data["Language"].map({"English": 0, "French": 1, "German": 2})
mapped_data
```

# Select the features

```python
x = mapped_data.iloc[:,1:4]
x
```

# Cluster the data

```python
kmeans = KMeans(2)
kmeans.fit(x)
```

# Get the cluster results

```python
identified_clusters = kmeans.fit_predict(x)
identified_clusters
```

# Data with clusters

```python
data_with_clusters = row_data.copy()
data_with_clusters['Cluster'] = identified_clusters
data_with_clusters
```

# Plot the clusters

```python
plt.scatter(data_with_clusters['Longitude'], data_with_clusters['Latitude'], c=data_with_clusters['Cluster'], cmap='rainbow')
plt.xlim(-180, 180)
plt.ylim(-90, 90)
plt.show()
```

# WCSS

```python
languages = mapped_data["Language"]
unique_language = languages.unique()
wcss = []
for i in range(1, unique_language.size):
    kmeans = KMeans(i)
    kmeans.fit(x)
    wcss.append(kmeans.inertia_)
wcss
```

```python
number_of_language = range(1, unique_language.size)
plt.plot(number_of_language, wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```
