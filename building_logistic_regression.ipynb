# Import the relevant libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
sns.set()
```

# Load the data from a .csv in the same folder

```python
row_data = pd.read_csv('admittance.csv')
row_data
```

# Change the column names to something more intuitive

```python
data = row_data.copy()
data['Admitted'] = data['Admitted'].map({'No': 0, 'Yes': 1})
data
```

# Declare the dependent and the independent variables

```python
y = data['Admitted'] # dependent variable
x1 = data['SAT'] # independent variable
```

# Regression
## Add a constant to the independent variable

```python
x = sm.add_constant(x1)
reg_logit = sm.Logit(y, x)
result_logit = reg_logit.fit()
result_logit.summary()
```

#

# Accuracy of the model

```python
result_logit.pred_table()
```

```python
cm_df = pd.DataFrame(result_logit.pred_table())
cm_df.columns = ['Predicted 0', 'Predicted 1']
cm_df = cm_df.rename(index={0: 'Actual 0', 1: 'Actual 1'})
cm_df
```

```python
cm = np.array(cm_df)
accuracy_train = (cm[0,0] + cm[1,1]) / cm.sum()
accuracy_train
```
