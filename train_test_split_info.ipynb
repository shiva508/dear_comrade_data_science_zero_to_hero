# Under fitting and Over fitting
## Under fitting: is when a model is too simple to capture the underlying patterns in the data, leading to poor performance on both training and test datasets.
## Over fitting, on the other hand, occurs when a model is too complex and captures noise in the training data as if it were a true pattern, resulting in excellent performance on the training data but poor generalization to new, unseen data.
## Over fitting can be mitigated through techniques such as cross-validation, regularization, splitting and pruning, while under fitting can be addressed by increasing model complexity or adding more relevant features.

# Over fitting can be minimized using train-test split
# split the dataset into a training set and a testing set. The training set is used to train the model, while the testing set is used to evaluate its performance on unseen data. This helps to ensure that the model generalizes well and does not overfit to the training data.

# Import the relevant libraries

```python
import numpy as np
from sklearn.model_selection import train_test_split
```

# Create a sample dataset

```python
data_set_one = np.arange(1, 101)
data_set_one
```

```python
data_set_two = np.arange(501, 601)
data_set_two
```

# Split the dataset into training and testing sets

```python
train_test_split(data_set_one)
```

# By default, the train_test_split function splits the data into 75% training and 25% testing sets. You can adjust this ratio using the test_size or train_size parameters.

```python
data_set_one_train, data_set_one_test= train_test_split(data_set_one, test_size=0.2, random_state=42)
```

# We can also split multiple datasets simultaneously, ensuring that the corresponding samples are kept together in the training and testing sets. This is particularly useful when you have features and labels that need to be split in a consistent manner.

```python
data_set_one_train, data_set_one_test, data_set_two_train, data_set_two_test = train_test_split(data_set_one, data_set_two, test_size=0.2, random_state=42)
```
