# Dictionaries
## Dictionaries store data in key-value pairs, allowing for efficient data retrieval based on unique keys.
## They are defined using curly braces {} and consist of keys and their corresponding values.
## Dictionaries are mutable, meaning their contents can be changed after creation.

```python
student_info = {
    'name': 'Alice',
    'age': 20,
    'courses': ['Math', 'Science', 'Art']
}
print(student_info)
```

```python
print(student_info['name'])
```

# Updating Dictionary Elements

```python
student_info["age"] = 34
print(student_info)
```

# Installing new key-value pairs

```python
team = {}
team['captain'] = 'Bob'
team['vice_captain'] = 'Charlie'
print(team)
```

# Different way to access dictionary values

```python
print(team.get("captain"))
print(team.get("keeper"))
```
