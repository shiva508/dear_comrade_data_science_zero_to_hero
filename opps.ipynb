# Classes
## Classes which enable bundling data and functionality together.
## Creating a new class creates a new type of object, allowing new instances of that type to be made.
## Each class instance can have attributes attached to it for maintaining its state.
## Class instances can also have methods (defined by its class) for modifying its state.

```python
class Opps:
    def greeting_message(self, name):
        return f"Hello {name},  welcome to Opps!"

opps_object = Opps()
print(opps_object.greeting_message("Shiva"))
```

```python
class ComplexNumber:
    def __init__(self, real, imaginary):
        self.real = real
        self.imaginary = imaginary

complexNumber_object = ComplexNumber(3, 4.0)
complexNumber_object.real,complexNumber_object.imaginary
```

#  Class and Instance Variables

```python
class Dog:
    species = "Canis familiaris"

    def __init__(self, name, age):
        self.name = name  # Instance variable
        self.age = age    # Instance variable
dog1 = Dog("Buddy", 5)
dog2 = Dog("Max", 3)
print(dog1.name, ",", dog1.age, ",", dog1.species)
print(dog2.name, ",", dog2.age, ",", dog2.species)
```
