# For Loop Information

```python
even_numbers = []
for num in range(22):
    if num % 2 == 0:
        even_numbers.append(num)
print("Even numbers from 0 to 20:", even_numbers)
```

```python
for event in even_numbers:
    print(event, end=" ")
```

```python
x = [0, 1, 2, 3, 4, 5]
for i in x:
    print(i , end=" ")
```

```python
for i in range(len(x)):
    print(x[i], end=" ")
```

```python
def count(limit):
    total = 0
    for num in range(limit):
        if num < 20:
            total += num
    return total
```

```python
sum_of_numbers = count(30)
print(sum_of_numbers)
```

```python

def count(input_list):
    total = 0
    for input_item in input_list:
        if input_item < 20:
            total+=1
    return total
nums = [1,35,12,24,31,51,70,100]
sum_of_numbers = count(nums)
print(sum_of_numbers)
```

```python
prices = {
    "apple": 0.5,
    "banana": 0.3,
    "orange": 0.8
}
quantity = {
    "apple": 4,
    "banana": 6,
    "orange": 3
}
for item in prices:
    print(f" Item: {item}, Price: ${prices[item]:.2f}, Quantity: {quantity[item]}")
    total_price = prices[item] * quantity[item]
    print(f"Total price for {item}: ${total_price:.2f}")
```

```python
prices = {
    "box_of_spaghetti" : 4,
    "lasagna"  : 5,
    "hamburger" : 2
   }
quantity = {
    "box_of_spaghetti" : 6,
    "lasagna"  : 10,
    "hamburger" : 0
    }

total_money_spent ={}
money_spent = 0
for food in prices:
    item_cost = prices[food] * quantity[food]
    total_money_spent[food] = item_cost
    money_spent += item_cost
print(f"Total money spent on food: ${money_spent}")
print("Breakdown of money spent on each food item:")
for food in total_money_spent:
    print(f"{food}: ${total_money_spent[food]}")
```

# for loop with enumerate()
## We use this when we need both index and value

```python
student_names = ["Shiva", "Dasari", "Gopala Swami" "Dasari"]
for index,student in enumerate(student_names):
    print(f" Index : {index}, name {student}")
```

# for loop with zip()

```python
student_names = ["Shiva", "Dasari", "Gopala Swami", "Dasari"]
student_rank  = [100, 99, 1, 2]
for rank, name in zip(student_rank,student_names):
    print(f" Rank : {rank}, name: {name}")
```

```python

import random

class InputValue:
    def __init__(self, value):
        self.value = value
    def __add__(self, other):
        return InputValue(self.value + other.value)
    def __sub__(self, other):
        return InputValue(self.value - other.value)
    def __mul__(self, other):
        return InputValue(self.value * other.value)
    def __truediv__(self, other):
        return InputValue(self.value / other.value)
    def __floordiv__(self, other):
        return InputValue(self.value // other.value)
    def __mod__(self, other):
        return InputValue(self.value % other.value)
    def __pow__(self, other):
        return InputValue(self.value ** other.value)
    def __lt__(self, other):
        return InputValue(self.value < other.value)
    def __gt__(self, other):
        return InputValue(self.value > other.value)
    def __le__(self, other):
        return InputValue(self.value <= other.value)
    def __ge__(self, other):
        return InputValue(self.value >= other.value)
    def __eq__(self, other):
        return InputValue(self.value == other.value)
    def __ne__(self, other):
        return InputValue(self.value != other.value)
    def __str__(self):
        return str(self.value)

values = [random.uniform(-1, 1) for _ in range(10)]
print(values)
input_values = [InputValue(random.uniform(0, 9)) for _ in range(10)]
for index, value in enumerate(input_values):
    print(f" Index : {index}, value: {value}")

for input_value_at, coefficient_value in zip(input_values, values):
    co_eff_input_value = InputValue(coefficient_value)
    input_value_at.__add__(co_eff_input_value)
    print(f" Coefficient : {coefficient_value}, value: {input_value_at}")
```

```python
act = sum(weight.value*co_eff for weight, co_eff in zip(input_values, values))
print(f"activation {act}")
```
