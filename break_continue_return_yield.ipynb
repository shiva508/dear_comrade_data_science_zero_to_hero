# return → giving full bucket of water
# yield → giving glass by glass when needed
# break → stopping the process immediately
# continue → skipping the current step and moving to the next one

# Return

```python
def get_numbers(num: int) -> list:
    numbers = []
    for i in range(1, num + 1):
        numbers.append(i)
    return numbers

result = get_numbers(10)
print(result)
```

# Yield

```python
def get_numbers_yield(num: int):
    for i in range(1, num + 1):
        yield i

results = get_numbers_yield(10)
for result in results:
    print(result)
```

```python
def start_end():
    print("Start")
    yield "Break - 1"
    print("Middle")
    yield "Break - 2"
    print("End")

results = start_end()
print(next(results))
print(next(results))
```

# Break

```python
for num in range(1, 11):
    if num == 5:
        print("Breaking the loop at number 5")
        break
    print(num)
```

# continue

```python
for num in range(1, 11):
    if num % 2 == 0:
        print(f"Skipping even number: {num}")
        continue
    print(num)
```
