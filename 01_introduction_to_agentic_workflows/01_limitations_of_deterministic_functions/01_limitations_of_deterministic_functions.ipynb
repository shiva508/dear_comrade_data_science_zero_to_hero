# Imports and setup

```python
from openai import OpenAI
```

# The Limitations of Deterministic Functions

```python
def traditional_generate_response(query):
    """
    A traditional function that returns predefined responses based on keywords.
    :param query:
    :return: string
    """
    query_lower = query.lower()
    if "weather" in query_lower:
        return "The weather today is sunny with a high of 75°F."
    elif "time" in query_lower:
        return "The current time is 2:00 PM."
    elif "hello" in query_lower or "hi" in query_lower:
        return "Hello! How can I help you today?"
    else:
        return "I'm not sure how to respond to that query."
result = traditional_generate_response("hello world")
print(result)
```

# OpenAI setup

```python
client = OpenAI(
    base_url="https://openai.vocareum.com/v1",
    api_key=""
)
```

# The Rise of LLM-Powered Agents

```python
def openai_generate_response(query):
    """
    Uses the OpenAI API to generate a response to the user's query.
    :param query:
    :return:
    """
    if not client:
        return "OpenAI client not initialized. Please set your API key."
    else:
        try:
            response = client.chat.completions.create(
            model="gpt-4",
            messages=[
                        {"role": "system", "content": "You are a helpful assistant."},
                        {"role": "user", "content": query}
                    ],
                max_tokens=150
                )
            return response.choices[0].message.content.strip()
        except Exception as e:
            return "OpenAI client not initialized. Please set your API key."
```

```python
response = openai_generate_response("What is the weather like today?")
print(response)
```
