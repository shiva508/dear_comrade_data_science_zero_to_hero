# Hard cored responses from a traditional function

```python
def get_hardcoded_answer(question):
    """
    Returns the hardcoded answer for the question
    :param question:
    :return:
    """
    question_lower = question.lower()
    if "What is a Gantt chart?".lower() in question_lower:
        return "A Gantt chart is a type of bar chart that illustrates a project schedule. It shows the start and finish dates of the various elements of a project, as well as their dependencies. Gantt charts are commonly used in project management to visualize the timeline of a project and track its progress."
    elif "Explain Agile.".lower() in question_lower:
        return "Agile is a project management and software development approach that emphasizes flexibility, collaboration, and customer satisfaction. It is based on the Agile Manifesto, which values individuals and interactions, working software, customer collaboration, and responding to change over following a strict plan. Agile methodologies, such as Scrum and Kanban, are commonly used to manage projects in a way that allows for iterative development and continuous improvement."
    elif "What is a sprint in Agile?".lower() in question_lower:
        return "A sprint in Agile is a time-boxed period, typically lasting 1-4 weeks, during which a specific set of work is completed and made ready for review. Sprints are a key component of the Scrum framework, where teams plan and execute their work in short cycles to allow for flexibility and adaptability. At the end of each sprint, the team reviews their progress and adjusts their plans for the next sprint based on feedback and changing requirements."
    elif "Define critical path.".lower() in question_lower:
        return "The critical path is the sequence of tasks in a project that determines the shortest possible duration to complete the project. It consists of tasks that have zero slack, meaning that any delay in these tasks will directly impact the overall project timeline. Identifying the critical path is essential for effective project management, as it helps prioritize tasks and allocate resources to ensure timely project completion."
    elif "What are project milestones?".lower() in question_lower:
        return "Project milestones are significant events or achievements within a project that mark the completion of a major phase or deliverable. They serve as checkpoints to track progress and ensure that the project is on schedule. Milestones can include the completion of key tasks, the approval of important documents, or the achievement of specific goals. They are often used in project management to help teams stay focused and motivated, as well as to communicate progress to stakeholders."
    else:
        return "I'm sorry, I only have information on a few specific program management topics. Could you ask about Gantt charts, Agile, sprints, critical path, or milestones?"
```

```python
response = get_hardcoded_answer("What is a Gantt chart?")
print(response)
```

# LLM agents can generate more dynamic and contextually relevant responses

## setup OpenAI client

```python
from openai import OpenAI
```

```python
client = OpenAI(
    base_url="https://openai.vocareum.com/v1",
    api_key=""
)
```

```python
def get_llm_answer(question):
    """
    Uses the OpenAI API to generate a response to the user's question.
    :param question:
    :return:
    """
    if not client:
        return "OpenAI client not initialized. Please set your API key."
    else:
        try:
            response = client.chat.completions.create(
            model="gpt-4",
            messages=[
                        {"role": "system", "content": "You are an expert assistant specializing in program management. Provide clear and concise answers to program management questions."},
                        {"role": "user", "content": question}
                    ],
                max_tokens=150
                )
            return response.choices[0].message.content.strip()
        except Exception as e:
            return "OpenAI client not initialized. Please set your API key."
```

```python
questions = [
    "What is a Gantt chart?",
    "Explain Agile.",
    "What is a sprint in Agile?",
    "Define critical path.",
    "What are project milestones?"
]
for question in questions:
    response = get_llm_answer(question)
    print(f"Question: {question}\nAnswer: {response}\n")
```

# Comparing the responses from the hardcoded function and the LLM agent

```python
def compare_answers(question):
    hardcoded_response = get_hardcoded_answer(question)
    llm_response = get_llm_answer(question)
    print(f"Question: {question}\n")
    print("-" * 50)
    print(f"Hardcoded Response:\n{hardcoded_response}\n")
    print(f"LLM Response:\n{llm_response}\n")
    print("=" * 50)
```

```python
questions = [
    "What is a Gantt chart?",
    "Explain Agile.",
    "What is a sprint in Agile?",
    "Define critical path.",
    "What are project milestones?"
]
for question in questions:
    compare_answers(question)
```
