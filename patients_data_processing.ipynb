# Patients Data Processing

```python
import pandas as pd
```

# Load the data from the provided CSV file

```python
raw_data = pd.read_csv('patients.csv')
# Display the first few rows of the dataset to understand its structure
raw_data.head()
```

# 1. Data Cleaning:

```python
raw_data["Diagnosis"] = raw_data["Diagnosis"].fillna("Unknown", inplace=True)
raw_data.drop_duplicates(inplace=True)
raw_data["Age"] = raw_data["Age"].astype(int)
raw_data = raw_data[raw_data["Age"] > 0]
raw_data["LabResult"] = raw_data["LabResult"].astype(int)
raw_data = raw_data[raw_data["LabResult"] > 0]
```

```python
data_encoding = {diagnosis: idx for idx, diagnosis in enumerate(raw_data["Diagnosis"].unique())}
raw_data["Diagnosis_Encoded"] = raw_data["Diagnosis"].map(data_encoding)
raw_data.head()
```
