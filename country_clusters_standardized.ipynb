# Cluster analysis (Heatmap and Dendrogram)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

# Load the data

```python
raw_data = pd.read_csv("country_clusters_standardized.csv", index_col="Country")
raw_data
```

```python
x_scaled = raw_data.copy()
x_scaled = x_scaled.drop(["Language"], axis=1)
x_scaled
```

# Plot the heatmap

```python
sns.clustermap(x_scaled, cmap="mako")
```
