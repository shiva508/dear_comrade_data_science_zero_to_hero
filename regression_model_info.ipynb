```python
import numpy as np
import pandas as pd
import scipy
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

# Load Dataset

```python
data = pd.read_csv('Simple_linear_regression.csv')
data.describe()
```

# Define independent and dependent variables

```python
y = data['GPA']
x1 = data['SAT']
```

# Explore the data

```python
plt.scatter(x1, y)
plt.xlabel('SAT', fontsize=20)
plt.ylabel('GPA', fontsize=20)
plt.show()
```

# Regression itself

```python
x = sm.add_constant(x1)
results = sm.OLS(y, x).fit()
results.summary()
```

```python
plt.scatter(x1, y)
# y = mx + b
yhat = x1 * 0.0017+ 0.2750
fig = plt.plot(x1, yhat, lw=4, c='orange', label='regression line')
plt.xlabel('SAT', fontsize=20)
plt.ylabel('GPA', fontsize=20)
plt.show()
```

## SST (Sum of Squares Total)
### SST = Square of the differences between the actual y(observed defendant variable) values and the mean of y
### Measure of total variability in the response variable
![SST](sst.png)
## SSR (Sum of Squares Regression)
### Sum of difference between the predicted y (predicted defendant variable) values and the mean of y
### SSR = Square of the differences between the predicted y (predicted defendant variable) values and the mean of y
![SSR](ssr.png)
## SSE (Sum of Squares Error)
### SSE = Square of the differences between the actual y (observed defendant variable) values and the predicted y (predicted defendant variable) values
### Measure of the unexplained variability by the regression model
![SSE](sse.png)
### SST = SSR + SSE
### R-squared = SSR / SST
### R-squared = Coefficient of Determination
### R-squared = 1 - (SSE / SST)
### OLS = Ordinary Least Squares(Stands for minimizing the sum of squared errors)
### R-squared = Explained variation / Total variation
### Adjusted R-squared = Adjusted for the number of predictors in the model(SSE and SST can be artificially reduced by adding more predictors to the model)
### Adjusted R-squared = 1 - [(1 - R-squared)(n - 1) / (n - k - 1)]
### n = number of observations
### k = number of predictors
