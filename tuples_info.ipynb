# Tuples
## Tuples are similar to lists but are immutable, meaning their elements cannot be changed after creation.
## Tuples are defined using parentheses () instead of square brackets [].
## They are often used to group related data that should not be modified.


```python
roll_numbers = (101, 102, 103, 104, 105)
print(roll_numbers)
```

```python
user_name = ('alice', 'bob', 'charlie')
print(user_name)
```

```python
SRH, MI= 'Sunrisers Hyderabad', 'Mumbai Indians'
```

```python
SRH
```

```python
student_info = [roll_numbers, user_name, SRH, MI]
print(student_info)
```

```python
(roll_no, name, team) = (101, 'alice', 'Sunrisers Hyderabad')
print(roll_no)
```

```python
(roll_no, name) = "101,alice".split(",")
print(roll_no)
print(name)
```

# Tuple with functions

```python
def ge_square_info(X):
    A = X ** 2
    P = 4 * X
    return A, P

area, perimeter = ge_square_info(5)
print("Area:", area)
print("Perimeter:", perimeter)
```
