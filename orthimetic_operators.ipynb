```python
import numpy as np
import tensorflow as tf
```

# Generating training data


```python
training_data = np.random.randint(1, 100, (1000, 2))
#test_data = training_data[:, 0] + training_data[:, 1]
test_data = np.sum(training_data, axis=1)
print(test_data.shape)
```

```python
hidden_layer_size = 80
output_layer_size = 1
arithmetic_model = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation=tf.nn.relu),
    tf.keras.layers.Dense(8, activation=tf.nn.relu),
    tf.keras.layers.Dense(1)
])
```

```python
arithmetic_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    loss=tf.keras.losses.mse,
    metrics=['mae']
)
```

```python
arithmetic_model.fit(
    x=training_data,
    y=test_data,
    epochs=100,
    verbose=2,
    batch_size=40,
)
```

```python
print(arithmetic_model.predict(np.array([[4,8]])))
```
