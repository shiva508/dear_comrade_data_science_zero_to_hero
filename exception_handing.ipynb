# Exception Handling

```python
10 * (10/0)
```

```python
try:
    10 * (10/0)
except ZeroDivisionError as e:
    print(f"Error: {e}")
```

```python
class B(Exception):
    pass
class C(B):
    pass
class D(C):
    pass

for cls in [B, C, D]:
    try:
        raise cls()
    except D:
        print("D")
    except C:
        print("C")
    except B:
        print("B")
```

```python
try:
    raise Exception('spam', 'eggs')
except Exception as inst:
    print(f"Caught an exception: {inst}")
    print(type(inst))    # the exception type
    print(inst.args)     # arguments stored in .args
    print(inst)          # __str__ allows args to be printed directly,
                         # but may be overridden in exception subclasses
    x, y = inst.args     # unpack args
    print('x =', x)
    print('y =', y)
```

```python
import os

try:
    f = open('myfile.txt')
    s = f.readline()
    i = int(s.strip())
except OSError as e:
    print(f"OS error: {e}")
except ValueError as e:
    print(f"Value error: {e}")
except Exception as e:
    print(f"Unexpected error: {e}")
```

```python
def divide_num_by_zero(num):
    return num / 0

try:
    divide_num_by_zero(0)
except ZeroDivisionError as e:
    print(f"Error: {e}")
```
