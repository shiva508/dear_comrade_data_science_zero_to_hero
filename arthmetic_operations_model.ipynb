# Load required imports

```python
import numpy as np
import tensorflow as tf
```

# Create training
## Step-1: Create train dataset which we will use to train on operations
## 87 + 50, 87 - 50, 87 * 50, 87 / 50
##### [[87 50]
##### [65 84]
##### ...
##### [74 54]
##### [72 86]
##### [72 41]]
## Step-2: create dataset to represent +,-,*,/, as we have different operations +,-,*,/.
## We can not directly use these, we use one-hot encoding
## Step-3: Combine Step-1, Step-2 to create actual training dataset

```python
num_of_samples = 21000
# Step - 1
train_dataset =  np.random.randint(1, 100, (num_of_samples, 2))
train_dataset = train_dataset
print(f"train dataset shape: {train_dataset.shape}")

# Step-2
operations_dateset = np.tile([0, 1, 2], 7000)
print(operations_dateset)
print(f"operation dateset shape: {operations_dateset.shape}")
# One hat encoding
one_hot_operations_dateset = np.eye(4)[operations_dateset]
print(one_hot_operations_dateset)
print(f"one_hot_operations_dateset shape: {one_hot_operations_dateset.shape}")

# Step -3
final_train_dataset = np.concatenate([train_dataset, one_hot_operations_dateset], axis=1)
print(f"final train_dataset shape: {final_train_dataset.shape}")
```

# Creating Test dataset
## we loop through number of samples and operations
## find the result and append to test dataset
## Convert to array

```python
test_dataset = []
addition_sample_count = 0
subtraction_sample_count = 0
multiplication_sample_count = 0
division_sample_count = 0
for i in range(num_of_samples):
    input_one, input_two = train_dataset[i]
    operation_type = operations_dateset[i]
    if operation_type == 0:
        test_dataset.append(input_one + input_two)
        addition_sample_count +=1
    elif operation_type == 1:
        test_dataset.append(input_one - input_two)
        subtraction_sample_count +=1
    else:
        test_dataset.append(input_one * input_two)
        multiplication_sample_count +=1

    # elif operation_type == 1:
    #     test_dataset.append(input_one - input_two)
    #     subtraction_sample_count +=1
    # elif operation_type == 2:
    #     test_dataset.append(input_one * input_two)
    #     multiplication_sample_count +=1
    # else:
    #     test_dataset.append(input_one / input_two)
    #     division_sample_count +=1

test_dataset = np.array(test_dataset)
test_dataset = test_dataset
print(f"test_dataset shape: {test_dataset.shape}")
print(f"addition sample count: {addition_sample_count}")
print(f"subtraction sample count: {subtraction_sample_count}")
print(f"multiplication sample count: {multiplication_sample_count}")
print(f"division sample count: {division_sample_count}")
```

# Creating model
## Input shape
### [input_one, input_two, +, -, *, /]
### 6x1 shape

```python
number_of_hidden_layers = 50
number_of_outputs = 1
math_operation_model = tf.keras.Sequential([
    tf.keras.layers.Dense(number_of_hidden_layers, activation=tf.nn.relu, input_shape=(6,)),
    tf.keras.layers.Dense(number_of_hidden_layers, activation=tf.nn.relu),
    tf.keras.layers.Dense(number_of_outputs),
])
```

# Adding optimizer and loss function

```python
math_operation_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
    loss=tf.keras.losses.mse,
    metrics=['mae']
)
```

# Training model

```python
epochs_size = 100
math_operation_model.fit(
    x=final_train_dataset,
    y=test_dataset,
    epochs=epochs_size,
    verbose=2,
    batch_size=40,
    shuffle=True
)
```

# Predict value

```python
math_operation_model.predict(np.array([[8, 4, 0, 0, 1, 0,]]))
```
