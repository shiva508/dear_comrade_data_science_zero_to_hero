```python
import random
def string_to_fract(input_string:str):
    if "/" in input_string:
        nd = input_string.split("/")
        n = float(nd[0])
        d = float(nd[1])
        ans = n/d
        return ans
    else:
        ans = float(input_string)
        return ans

def one_step_mul():
    a = random.randint(1, 11)
    b = random.randint(2, 24)
    print(a, " * x=", b)
    ans_in = (input("x= "))
    gene_answer = a/b
    if string_to_fract(ans_in) == gene_answer:
        print("Correct answer")
    else:
        print("Try again ")
        print("Correct answer is ", gene_answer)
def one_step_add():
    a = random.randint(-4, 11)
    b = random.randint(2, 24)
    print("x + ",a, " = ", b)
    ans_in = (input("x= "))
    gen_answer = b-a
    if ans_in == gen_answer:
        print("Correct answer")
    else:
        print("Try again ")
        print("Correct answer is ", gen_answer)
one_step_mul()
one_step_add()
```
