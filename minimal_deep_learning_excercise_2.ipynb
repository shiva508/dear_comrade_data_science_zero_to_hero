```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

# define total number of observations
# xs of 1000x1
# zs of 1000x1
# create inputs using xs,zs of size 1000x2
# generate noise
# generate targets=11*xs-5zs+10+noise
# Reshape target before plot and reshape back to original
# plot xs,zs,targets
# define init_value
# using init_value generate weights
# using init_value generate biases
# define learning rate
# y = x*w + bias
# loop for 100
# generate outputs
# generate deltas
# fine lose
# scale deltas
# find revised weights and biases


```python
observations = 2000
xs = np.random.uniform(low=-10, high=10, size=(observations,1))
zs = np.random.uniform(low=-10, high=10, size=(observations,1))
inputs = np.column_stack((xs,zs))
print(f"shape of xs {xs.shape}")
print(f"shape of zs {zs.shape}")
print(f"shape of inputs {inputs.shape}")
```

```python
noise = np.random.uniform(low=-1, high=1, size=(observations,1))
print(f" shape of noise {noise.shape}")
```

```python
targets = 11 * xs + 5*zs - 10 + noise
print(f" shape of targets {targets.shape}")
targets = targets.reshape(observations,)
print(f"reshaped {targets.shape}")
```

```python

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot(xs, zs, targets)
ax.view_init(azim=100)
plt.show()
```

```python
targets = targets.reshape(observations,1)
print(f"reshaped {targets.shape}")
```

```python
init_value = 0.1
#init_range = 0.1
weight = np.random.uniform(low=-init_value, high=init_value, size=(2,1))
biases = np.random.uniform(low=-init_value, high=init_value, size=1)
```

```python
learning_rate = 0.02
#learning_rate = 0.02
for i in range(100):
#for i in range(100):
    outputs = np.dot(inputs, weight) + biases
    deltas = outputs - targets
    loses = np.sum(deltas**2)/2/observations
    print(loses)
    delta_scaled = deltas/observations
    weight = weight - learning_rate * np.dot(inputs.T, delta_scaled)
    biases = biases - learning_rate * np.sum(delta_scaled)
```

```python
print(weight)
print(biases)
```

```python
plt.plot(outputs, targets)
plt.xlabel("Inputs")
plt.ylabel("Targets")
plt.show()
```
