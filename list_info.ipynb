# List Information
##

```python
user_names = ['alice', 'bob', 'charlie', 'david', 'eve']
```

```python
print(user_names[1])
print(user_names[:2])
print(user_names[2:])
print(user_names[1:2])
```

# Updating List Elements

```python
user_names[0] = 'alex'
print(user_names)
```

# Deleting List Elements

```python
del user_names[2]
print(user_names)
```

# Adding Elements to a List

```python
user_names.append("frank")
print(user_names)
```

# Adding multiple elements using extend()

```python
user_names.extend(["grace", "heidi"])
print(user_names)
```

# Slicing Lists

```python
print(user_names[1:4])
print(user_names[-3:])
print(user_names[:-3])
```

# Index of an Element

```python
user_names.index("david")
```

# Combining Lists

```python
section_one = ['alice', 'bob', 'charlie']
section_two = ['david', 'eve', 'frank']
combined_sections = section_one + section_two
print(combined_sections)
```

```python
section_one = ['alice', 'bob', 'charlie']
section_two = ['david', 'eve', 'frank']
combined_sections = [section_one , section_two]
print(combined_sections)
```

# Sorting a List

```python
section_one = ['alice', 'charlie', 'bob']
print(section_one)
section_one.sort()
print(section_one)
```

```python
section_one = ['alice', 'charlie', 'bob']
print(section_one)
section_one.sort(reverse=True)
print(section_one)
```
